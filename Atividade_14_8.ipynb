```python
nota=float(input(f"digite sua nota."))
if nota >+7:
  print(f"parabens, você tirou {nota} você esta aprovado!")
elif nota >=5:
  print(f"quase lá, voce tirou {nota}, você esta de recuperação.")
else:
  print(f"que decepissão, você tirou {nota}, você está reprovado.")
```

```python
print("não digite 'kg' nem 'm' nos valores abaixo.")
altura=float(input(f"digite sua altura(ex:1.75):"))
peso=float(input(f"digite seu peso(ex:65):"))
IMC= peso/(altura*altura)
print(f"seu IMC é {IMC}, ou seja...")
if IMC >=30:
  print(f"voce é obeso")
elif IMC >=25:
  print(f"voce é sobrepeso")
elif IMC >=18.5:
  print(f"voce tem o peso normal")
else:
  print(f"voce está abaixo do peso")
```
